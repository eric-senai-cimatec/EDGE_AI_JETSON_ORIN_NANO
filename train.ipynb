```python
from ultralytics import YOLO
```

```python
# Load a model
model = YOLO("yolo26n.pt")  # load a pretrained model (recommended for training)
```

```python
# Train the model
result = model.train(data="construction-ppe.yaml", epochs=100, imgsz=640, batch=16, device=0)
```

```python
# Carregar o modelo treinado
model = YOLO(r"runs\detect\train\weights\best.pt")
```

```python
# Avaliar especificamente no conjunto de TESTE
metrics = model.val(data="construction-ppe.yaml", split='test')
```

```python
import os
from PIL import Image
import cv2

# construction-ppe\images\test

# Path to test images
test_path = r"construction-ppe\images\test"

# Get all image files
image_files = sorted([f for f in os.listdir(test_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])

# Run inference and collect results
results = []
images = []

for img_file in image_files:
    img_path = os.path.join(test_path, img_file)
    result = model.predict(img_path, conf=0.5)
    
    # Get annotated image
    annotated_img = result[0].plot()
    images.append(Image.fromarray(annotated_img[:, :, ::-1]))

# Create GIF with 3 seconds per frame (duration in milliseconds)
if images:
    images[0].save(
        'predictions.gif',
        save_all=True,
        append_images=images[1:],
        duration=3000,
        loop=0
    )
    print(f"GIF created successfully with {len(images)} frames!")
```
